### 1. Import Library

```python
import pandas as pd

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score
)
```

### 2. Load Dataset

```python
# Membuat Data Aktual dan Prediksi
aktual = [
    'Lulus',
    'Lulus',
    'Lulus',
    'Lulus',
    'Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
]

prediksi = [
    'Lulus',
    'Lulus',
    'Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
    'Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
]
```

```python
# Membuat Confusion Matrix
cm = confusion_matrix(
    aktual,
    prediksi,
    labels=['Lulus', 'Tidak Lulus']
)

cm
```

```python
# Menampilkan Confusion Matrix dalam bentuk Tabel
cm_df = pd.DataFrame(
    cm,
    index=[
        'Aktual Lulus',
        'Aktual Tidak Lulus'
    ],
    columns=[
        'Prediksi Lulus',
        'Prediksi Tidak Lulus'
    ]
)

cm_df
```

```python
# Menghitung accuracy
accuracy = accuracy_score(
    aktual,
    prediksi
)

print(
    f"Accuracy : {accuracy*100:.2f}%"
)
```

```python
# Menghitung Precision
precision = precision_score(
    aktual,
    prediksi,
    pos_label='Lulus'
)

print(
    f"Precision : {precision*100:.2f}%"
)
```

```python
# Menghitung Recall
recall = recall_score(
    aktual,
    prediksi,
    pos_label='Lulus'
)

print(
    f"Recall : {recall*100:.2f}%"
)
```
